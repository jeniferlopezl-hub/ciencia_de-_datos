```
class Persona:
    def __init__(self, nombre, edad):
        self.nombre = nombre
        self.edad = edad

    def mostrar_info(self):
        print("Nombre:", self.nombre)
        print("Edad:", self.edad)


p1 = Persona("Juan", 25)
p1.mostrar_info()
```

```
class Medicamento:
    def __init__(self, nombre, stock):
        self._stock = stock

    def vender(self, cantidad):
        if cantidad <= self._stock:
            self._stock -= cantidad
            print("Venta realizada. Stock:", self._stock)
        else:
            print("No hay suficiente stock")


med = Medicamento("Paracetamol", 100)
med.vender(20)
```

```
class MedicamentoBase:
    def __init__(self, nombre):
        self.nombre = nombre

    def mostrar(self):
        print("Medicamento:", self.nombre)


class Antibiotico(MedicamentoBase):
    def mostrar(self):
        print("Antibiótico:", self.nombre)


m = MedicamentoBase("Genérico")
a = Antibiotico("Amoxicilina")

m.mostrar()
a.mostrar()
```

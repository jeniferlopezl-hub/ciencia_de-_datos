```
#programacion de objetos

class Medicamento:
    def __init__(self, nombre, dosis_mg, stock):
        self.nombre = nombre
        self.dosis_mg = dosis_mg
        self.stock = stock

    def mostrar_info(self):
        print("Medicamento:", self.nombre)
        print("Dosis:", self.dosis_mg, "mg")
        print("Stock:", self.stock)

    def vender(self, cantidad):
        if cantidad <= self.stock:
            self.stock -= cantidad
            print("Venta realizada. Stock restante:", self.stock)
        else:
            print("No hay suficiente stock.")


# Uso
med1 = Medicamento("Paracetamol", 500, 100)

med1.mostrar_info()
med1.vender(20)
med1.vender(90)
```
